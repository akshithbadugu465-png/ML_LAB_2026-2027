**2. Cell 2 – Import libraries**

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

np.random.seed(42)
rng = np.random.default_rng(42)

print("All libraries imported successfully!")
```

**3. Cell 3 – Create Dataset 1 (Student Performance)**
Algorithm: Linear Regression.

Objective: Predict student marks based on study hours.

```python

n = 300

study_hours = rng.uniform(1, 10, n)

marks = (
    12 + 8.2 * study_hours
    + rng.normal(0, 6, n)
)

student_data = pd.DataFrame({
    "Study_Hours": study_hours,
    "Marks": marks
})

print("STUDENT PERFORMANCE DATASET")
display(student_data.head(10))

print("Dataset Shape:", student_data.shape)
```

**4. Cell 4 – Train Linear Regression**

```python

X1 = student_data[["Study_Hours"]]
y1 = student_data["Marks"]

X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y1, test_size=0.2, random_state=42
)

linear_model = LinearRegression()

linear_model.fit(X1_train, y1_train)

linear_pred = linear_model.predict(X1_test)

linear_r2 = r2_score(y1_test, linear_pred)
linear_mae = mean_absolute_error(y1_test, linear_pred)
linear_rmse = np.sqrt(mean_squared_error(y1_test, linear_pred))

print("LINEAR REGRESSION RESULTS")
print("Intercept:", linear_model.intercept_)
print("Coefficient:", linear_model.coef_[0])
print("R2 Score:", linear_r2)
print("MAE:", linear_mae)
print("RMSE:", linear_rmse)
```

**5. Cell 5 – Create Dataset 2 (House Price Prediction)**
Algorithm: Multiple Linear Regression.

Objective: Predict house prices using area, bedrooms and house age.

```python

area = rng.uniform(600, 3500, n)
bedrooms = rng.integers(1, 6, n)
house_age = rng.uniform(0, 40, n)

price = (
    50000
    + 145 * area
    + 18000 * bedrooms
    - 1200 * house_age
    + rng.normal(0, 25000, n)
)

house_data = pd.DataFrame({
    "Area": area,
    "Bedrooms": bedrooms,
    "House_Age": house_age,
    "Price": price
})

print("HOUSE PRICE DATASET")
display(house_data.head(10))

print("Dataset Shape:", house_data.shape)
```

**6. Cell 6 – Train Multiple Linear Regression**

```python

X2 = house_data[["Area", "Bedrooms", "House_Age"]]
y2 = house_data["Price"]

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42
)

multiple_model = LinearRegression()

multiple_model.fit(X2_train, y2_train)

multiple_pred = multiple_model.predict(X2_test)

multiple_r2 = r2_score(y2_test, multiple_pred)
multiple_mae = mean_absolute_error(y2_test, multiple_pred)
multiple_rmse = np.sqrt(
    mean_squared_error(y2_test, multiple_pred)
)

print("MULTIPLE LINEAR REGRESSION RESULTS")
print("Intercept:", multiple_model.intercept_)
print("Coefficients:", multiple_model.coef_)
print("R2 Score:", multiple_r2)
print("MAE:", multiple_mae)
print("RMSE:", multiple_rmse)
```

**7. Cell 7 – Create Dataset 3 (Customer Purchase Prediction)**
Algorithm: Logistic Regression.

Objective: Predict whether a customer purchases a product or not.

```python

age = rng.integers(18, 66, n)
salary = rng.uniform(15000, 150000, n)
browsing_time = rng.uniform(1, 60, n)

z = (
    -4.0
    + 0.035 * (age - 18)
    + 0.000035 * (salary - 15000)
    + 0.055 * browsing_time
)

probability = 1 / (1 + np.exp(-z))

purchase = rng.binomial(1, probability)

customer_data = pd.DataFrame({
    "Age": age,
    "Salary": salary,
    "Browsing_Time": browsing_time,
    "Purchase": purchase
})

print("CUSTOMER PURCHASE DATASET")
display(customer_data.head(10))

print("Dataset Shape:", customer_data.shape)
print("\nPurchase distribution:")
print(customer_data["Purchase"].value_counts())
```

**8. Cell 8 – Train Logistic Regression**

```python

X3 = customer_data[
    ["Age", "Salary", "Browsing_Time"]
]

y3 = customer_data["Purchase"]

X3_train, X3_test, y3_train, y3_test = train_test_split(
    X3, y3,
    test_size=0.2,
    random_state=42,
    stratify=y3
)

logistic_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

logistic_model.fit(X3_train, y3_train)

logistic_pred = logistic_model.predict(X3_test)

accuracy = accuracy_score(y3_test, logistic_pred)
precision = precision_score(y3_test, logistic_pred)
recall = recall_score(y3_test, logistic_pred)
f1 = f1_score(y3_test, logistic_pred)

print("LOGISTIC REGRESSION RESULTS")
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

**9. Cell 9 – Combined performance comparison**
This cell combines the results of all three algorithms into one report.

```python

results = pd.DataFrame({
    "Algorithm": [
        "Linear Regression",
        "Multiple Linear Regression",
        "Logistic Regression"
    ],
    "Evaluation Metric": [
        "R2 Score",
        "R2 Score",
        "Accuracy"
    ],
    "Score": [
        linear_r2,
        multiple_r2,
        accuracy
    ]
})

print("========== FINAL ML RESULTS ==========")

display(results)

print("\nDetailed Regression Results")

print("Linear Regression:")
print("MAE:", linear_mae, "RMSE:", linear_rmse)

print("\nMultiple Linear Regression:")
print("MAE:", multiple_mae, "RMSE:", multiple_rmse)

print("\nLogistic Regression:")
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

**10. Cell 10 – Matplotlib visualization**
This cell generates four graphs to show the overall outcome.

```python

fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# Graph 1: Linear Regression

axes[0, 0].scatter(
    study_hours, marks,
    alpha=0.6,
    label="Student Data"
)

line_x = np.linspace(
    study_hours.min(),
    study_hours.max(),
    100
)

line_y = (
    linear_model.intercept_
    + linear_model.coef_[0] * line_x
)

axes[0, 0].plot(
    line_x, line_y,
    label="Regression Line"
)

axes[0, 0].set_title("Student Marks Prediction")
axes[0, 0].set_xlabel("Study Hours")
axes[0, 0].set_ylabel("Marks")
axes[0, 0].legend()

# Graph 2: Multiple Linear Regression

axes[0, 1].scatter(
    y2_test, multiple_pred,
    alpha=0.6
)

minimum = min(y2_test.min(), multiple_pred.min())
maximum = max(y2_test.max(), multiple_pred.max())

axes[0, 1].plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

axes[0, 1].set_title("House Price: Actual vs Predicted")
axes[0, 1].set_xlabel("Actual Price")
axes[0, 1].set_ylabel("Predicted Price")

# Graph 3: Logistic Regression

cm = confusion_matrix(y3_test, logistic_pred)

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Purchase", "Purchase"]
).plot(ax=axes[1, 0], colorbar=False)

axes[1, 0].set_title("Customer Purchase Prediction")

# Graph 4: Combined model scores

axes[1, 1].bar(
    results["Algorithm"],
    results["Score"]
)

axes[1, 1].set_title("ML Model Score Comparison")
axes[1, 1].set_ylabel("Score")
axes[1, 1].set_ylim(0, 1.05)
axes[1, 1].tick_params(
    axis="x",
    rotation=15
)

plt.tight_layout()

plt.savefig(
    "ML_Comparative_Analysis.png",
    dpi=300
)

plt.show()
```
